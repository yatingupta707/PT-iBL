```python
import pandas as pd
```

```python
cols = ['id','val_high','p_high','val_low','val_safe','sure','d1','mode']
est  = pd.read_csv('60competitionset.dat',   sep=r'\s+', header=None, names=cols)
```

```python
est_final = est[['id','val_high','p_high','val_low','val_safe']].copy()
est_final.head()
```

```python
updated_records_est=[]
for _,row in est_final.iterrows():
    updated_records_est.append({
        'id': row['id'],
        'val_high': row['val_high'],
        'p_high': row['p_high'],
        'val_low': row['val_low'],
        'p_low': 1-row['p_high'],
        'val_safe': row['val_safe'],
        'p_safe': 1
    })

updated_est_df = pd.DataFrame(updated_records_est)
updated_est_df.head()
```

```python
updated_est_df.to_csv('TPTcompetitionSet.csv', index=False)
```
